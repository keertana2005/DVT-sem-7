```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Fill missing values with the mean
df["Temperature"] = df["Temperature"].fillna(df["Temperature"].mean())
df["Fuel_Price"] = df["Fuel_Price"].fillna(df["Fuel_Price"].mean())
df["CPI"] = df["CPI"].fillna(df["CPI"].mean())
df["Unemployment"] = df["Unemployment"].fillna(df["Unemployment"].mean())
df["Weekly_Sales"] = df["Weekly_Sales"].fillna(df["Weekly_Sales"].mean())

# Display the updated dataset
print(df)
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Fill missing values in all numeric columns with their mean
numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# Display the dataset
print(df)
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Select numerical columns
numeric_cols = df.select_dtypes(include='number').columns

# Calculate IQR and Outlier Limits for each numerical column
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    print(f"\nColumn: {col}")
    print(f"Q1           : {Q1}")
    print(f"Q3           : {Q3}")
    print(f"IQR          : {IQR}")
    print(f"Lower Limit  : {lower_limit}")
    print(f"Upper Limit  : {upper_limit}")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Select numerical columns
numeric_cols = df.select_dtypes(include='number').columns

# Count outliers using IQR
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"{col}: {len(outliers)} outliers")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Select numerical columns
numeric_cols = df.select_dtypes(include='number').columns

# Apply Min-Max Normalization
df[numeric_cols] = (df[numeric_cols] - df[numeric_cols].min()) / (
    df[numeric_cols].max() - df[numeric_cols].min()
)

# Display normalized dataset
print(df)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Min-Max Normalization for Weekly_Sales
df["Weekly_Sales"] = (
    df["Weekly_Sales"] - df["Weekly_Sales"].min()
) / (
    df["Weekly_Sales"].max() - df["Weekly_Sales"].min()
)

print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Create LabelEncoder object
le = LabelEncoder()

# Apply Label Encoding to the Date column
df["Date"] = le.fit_transform(df["Date"])

# Display the encoded dataset
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Load the dataset
df = pd.read_csv("/kaggle/input/datasets/yasserh/walmart-dataset/Walmart.csv")

# Create LabelEncoder object
le = LabelEncoder()

# Find categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Apply Label Encoding
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

# Display the encoded dataset
print(df)
```
